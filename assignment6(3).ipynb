```python
class Account:
    def __init__(self, account_number, customer_name, balance):
        self.account_number = account_number
        self.customer_name = customer_name
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(f"Rs. {amount} deposited. New balance: {self.balance}")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Rs. {amount} withdrawn. New balance: {self.balance}")
        else:
            print("Insufficient balance!")
    def calculate_profit(self):
        print("No profit calculation for this account type.")
    def show_info(self):
        print("Account Number:", self.account_number)
        print("Customer Name:", self.customer_name)
        print("Balance:", self.balance)
class SavingsAccount(Account):
    def calculate_profit(self):
        profit = self.balance * 0.05
        print(f"Savings Account Profit: Rs. {profit}")
class CurrentAccount(Account):
    def calculate_profit(self):
        print("Current Account: No profit is given.")
class BusinessAccount(Account):
    def calculate_profit(self):
        profit = self.balance * 0.08  #
        print(f"Business Account Profit: Rs. {profit}")
s1 = SavingsAccount("SA001", "Ali Raza", 100000)
c1 = CurrentAccount("CA001", "Sara Khan", 200000)
b1 = BusinessAccount("BA001", "Hamza Traders", 500000)
s1.show_info()
s1.deposit(5000)
s1.withdraw(2000)
s1.calculate_profit()
print("------------")
c1.show_info()
c1.deposit(10000)
c1.calculate_profit()
print("------------")
b1.show_info()
b1.withdraw(50000)
b1.calculate_profit()
print("------------")
```
